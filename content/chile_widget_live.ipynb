# Scenarios with sliders

The Chile macro-fiscal model with sliders for five shocks: the VAT rate, pension spending, the
health-spending elasticity, female participation and the copper price. Move one or more sliders,
give the scenario a name and press **Run scenario**; the solution is kept and shown next to the
baseline and the scenarios run before it.

Python runs inside your browser; nothing is installed on your computer. The first cell installs
ModelFlow into the browser and takes a little while; after a reload it has to run again.
In the book, start Python with the power button and run the cells from the top. In JupyterLite
use **Run > Run All Cells**.

The scenarios and the sliders are explained in the book
[ModelFlow models for Chile](https://ibhansen.github.io/chile/).

```python
import sys

if sys.platform == 'emscripten':  # in the browser: JupyterLite or the book's live code (thebe-lite)
    import micropip
    await micropip.install('modelflowib>=2.87')
    # the book's live code has no files, so the model and the slider definitions are fetched
    # from the chile_lite repo (JupyterLite has them already)
    from pathlib import Path
    from pyodide.http import pyfetch
    for file in ['data/chilepens.pcim', 'chile_widgets.py']:
        if not Path(file).exists():
            Path(file).parent.mkdir(exist_ok=True)
            response = await pyfetch(f'https://raw.githubusercontent.com/IbHansen/chile_lite/main/content/{file}')
            Path(file).write_bytes(await response.bytes())
    sys.path.insert(0, '')        # the notebook's folder, for chile_widgets.py
```
